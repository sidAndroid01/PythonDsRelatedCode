```python
import numpy as np
## y=8x+2
x=np.array([1,4,5], dtype=float)
y=np.array([10,34,42], dtype=float)
n=len(x)

m=0.0
c=0.0
# epochs=1
learingRate=0.01
epochs = 1000

for epoch in range(epochs):
    ## forward pass
    yPred=m*x + c

    ## mse loss
    loss = np.mean((yPred-y)**2)

    ## dloss/da
    dm = (2/n)*np.sum((yPred-y)*x)
    dc = (2/n)*np.sum(yPred-y)

    m = m - (dm*learingRate)
    c = c - (dc*learingRate)

    print(f"Epoch : {epoch}")
    print(f"Epoch : {epoch}, curr losss {loss:0.6f}")

print(f"Final m is {m:0.6f}")
print(f"Final c is {c:0.6f}")


def accuracy(yPred, yTrue):
    n=len(yPred)
    correct=np.sum(yPred==yTrue)
    return correct/n

yPred=np.array([1,2,3,2,1])
yTrue=np.array([1,1,2,3,1])

print(f"Accuracy is {accuracy(yPred, yTrue)}")
```

```python
import torch

x=torch.tensor([1,2,3], dtype=torch.float32)
y=torch.tensor([12,20,28], dtype=torch.float32)
## y= 8x+4(y=mx+c)
m=torch.tensor(0.0, dtype=torch.float32, requires_grad=True)
c=torch.tensor(0.0, dtype=torch.float32, requires_grad=True)

epochs=1000
learningRate=0.01
for i in range(epochs):
    ## forward pass
    yPred=m*x + c

    ## mse loss
    loss = torch.mean((yPred-y)**2)

    ## backward pass
    loss.backward()

    ## update weights
    with torch.no_grad():
        m -= learningRate*m.grad
        c -= learningRate*c.grad

        ## zero the gradients after updating weights
        m.grad.zero_()
        c.grad.zero_()

    print(f"Epoch : {i}, curr losss {loss:0.6f}")

print(f"Final m is {m:0.6f}")
print(f"Final c is {c:0.6f}")
```

### Notes

- `np.random.seed(0)` fixes the random sequence but each call to a random function consumes values from that sequence. So `matrix` and `matrix1` will be different, but reproducible across runs.
- Use `np.array(...)` (or `np.asarray(...)`) to create arrays from Python lists. `np.ndarray(...)` is a low-level constructor that expects a `shape` argument and raw uninitialized memory; passing a Python list to it will raise an error or behave unexpectedly.

Examples:

- Recommended: `matrix2 = np.array([[1,2,3],[1,2,3]])`  # works
- Avoid: `matrix2 = np.ndarray([[1,2,3],[1,2,3]])`  # incorrect usage

import torch
import torch.nn as nn
import torch.optim as optim

from torchvision import datasets, transforms
from torch.utils.data import DataLoader


# -----------------------------
# 1. Load and preprocess MNIST
# -----------------------------

transform = transforms.ToTensor()

train_dataset = datasets.MNIST(
    root="./data",
    train=True,
    download=True,
    transform=transform
)

test_dataset = datasets.MNIST(
    root="./data",
    train=False,
    download=True,
    transform=transform
)


# -----------------------------
# 2. Create batches
# -----------------------------

train_loader = DataLoader(
    train_dataset,
    batch_size=64,
    shuffle=True
)

test_loader = DataLoader(
    test_dataset,
    batch_size=64,
    shuffle=False
)


# -----------------------------
# 3. Define CNN
# -----------------------------

model = nn.Sequential(

    nn.Conv2d(
        in_channels=1,
        out_channels=16,
        kernel_size=3,
        padding=1
    ),

    nn.ReLU(),

    nn.MaxPool2d(
        kernel_size=2,
        stride=2
    ),

    nn.Conv2d(
        in_channels=16,
        out_channels=32,
        kernel_size=3,
        padding=1
    ),

    nn.ReLU(),

    nn.MaxPool2d(
        kernel_size=2,
        stride=2
    ),

    nn.Flatten(),

    nn.Linear(
        32 * 7 * 7,
        10
    )
)


# -----------------------------
# 4. Loss + optimizer
# -----------------------------

criterion = nn.CrossEntropyLoss()

optimizer = optim.Adam(
    model.parameters(),
    lr=0.001
)


# -----------------------------
# 5. Training
# -----------------------------

for epoch in range(5):

    model.train()

    for images, labels in train_loader:

        # Forward pass
        outputs = model(images)

        # Calculate loss
        loss = criterion(outputs, labels)

        # Clear previous gradients
        optimizer.zero_grad()

        # Backpropagation
        loss.backward()

        # Update weights
        optimizer.step()

    print(
        f"Epoch {epoch + 1}, "
        f"Loss: {loss.item():.4f}"
    )


# -----------------------------
# 6. Testing
# -----------------------------

model.eval()

correct = 0
total = 0

with torch.no_grad():

    for images, labels in test_loader:

        outputs = model(images)

        predictions = outputs.argmax(dim=1)

        total += labels.size(0)

        correct += (
            predictions == labels
        ).sum().item()


accuracy = correct / total

print(
    f"Test Accuracy: {accuracy * 100:.2f}%"
)
