# Cost Functions in Neural Networks

## Quadratic Cost (Mean Squared Error - MSE)
- **Formula:** C = Σ(actual - predicted)² / n
- **Use case:** Regression problems
- **Pros:** Simple, differentiable, easy to optimize
- **Cons:** Sensitive to outliers, can saturate gradients

## Cross Entropy Cost
- **Formula:** C = -Σ[y * log(ŷ) + (1-y) * log(1-ŷ)]
- **Use case:** Classification problems (especially binary classification)
- **Pros:** Better gradient flow, avoids saturation, naturally suits probability outputs
- **Cons:** Requires normalized inputs (0-1 range)

## Key Differences
| Aspect | Quadratic | Cross Entropy |
|--------|-----------|---------------|
| **Best For** | Regression | Classification |
| **Output** | Continuous values | Probabilities |
| **Gradient** | Can vanish | More stable |
| **Outlier Sensitivity** | High | Lower |

````python
import matplotlib.pyplot as plt
import numpy as np
%matplotlib inline

def sigMoid(z):
    """
    Sigmoid activation function
    Formula: σ(z) = 1 / (1 + e^(-z))
    Squashes any input to a value between 0 and 1
    """
    return 1/(1+np.exp(-z))

# np.linspace creates 100 evenly spaced values from -10 to 10
sample_z = np.linspace(-10, 10, 100)
print(f"Sample Z values (first 5): {sample_z[:5]}")
print(f"Total values: {len(sample_z)}")

# When you pass the array to sigmoid, NumPy applies it to EACH value
# This is called "vectorization" or "element-wise operation"
sigmoid_values = sigMoid(sample_z)
print(f"\nSigmoid values (first 5): {sigmoid_values[:5]}")

# Visualize the sigmoid curve
plt.figure(figsize=(10, 6))
plt.plot(sample_z, sigmoid_values, linewidth=2, color='blue')
plt.grid(True, alpha=0.3)
plt.xlabel('Z (Input)', fontsize=12)
plt.ylabel('Sigmoid(Z) (Output)', fontsize=12)
plt.title('Sigmoid Activation Function', fontsize=14)
plt.axhline(y=0.5, color='red', linestyle='--', alpha=0.5, label='y=0.5')
plt.axvline(x=0, color='green', linestyle='--', alpha=0.5, label='z=0')
plt.legend()
plt.show()

print("\nKey observations:")
print(f"- When z=-10: sigmoid ≈ {sigMoid(-10):.6f} (very close to 0)")
print(f"- When z=0: sigmoid = {sigMoid(0):.6f} (exactly 0.5)")
print(f"- When z=10: sigmoid ≈ {sigMoid(10):.6f} (very close to 1)")
````

````python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs

# ── Step 1: Generate two clusters of fake data ──────────────────────────
# make_blobs creates labelled clusters — perfect for classification demos
features, labels = make_blobs(
    n_samples=50,        # 50 total points
    n_features=2,        # 2 features (x and y position)
    centers=2,           # 2 clusters (class 0 and class 1)
    cluster_std=1.0,     # how spread out each cluster is
    random_state=42      # fixed seed so you get same result every run
)

# features shape: (50, 2)  → 50 rows, 2 columns
# labels shape:   (50,)    → 50 values, each 0 or 1

print("Features shape:", features.shape)
print("Labels shape:  ", labels.shape)
print("First 3 points:", features[:3])
print("First 3 labels:", labels[:3])

# ── Step 2: Calculate the optimal decision boundary ─────────────────────
# Find the centroid (center) of each cluster
centroid_0 = features[labels == 0].mean(axis=0)
centroid_1 = features[labels == 1].mean(axis=0)

print(f"\nCluster 0 centroid: {centroid_0}")
print(f"Cluster 1 centroid: {centroid_1}")

# The decision boundary passes through the midpoint between centroids
midpoint = (centroid_0 + centroid_1) / 2
print(f"Midpoint: {midpoint}")

# Calculate the slope perpendicular to the line connecting centroids
# Slope of line between centroids
if centroid_1[0] - centroid_0[0] != 0:
    slope_connecting = (centroid_1[1] - centroid_0[1]) / (centroid_1[0] - centroid_0[0])
    # Perpendicular slope (negative reciprocal)
    boundary_slope = -1 / slope_connecting
else:
    boundary_slope = 0

# y - y1 = m(x - x1) → y = mx - mx1 + y1
# y = boundary_slope * x + (midpoint[1] - boundary_slope * midpoint[0])
boundary_intercept = midpoint[1] - boundary_slope * midpoint[0]

print(f"Decision boundary: y = {boundary_slope:.2f}x + {boundary_intercept:.2f}")

# ── Step 3: Create x values for the decision boundary line ──────────────
x = np.linspace(features[:, 0].min() - 1, features[:, 0].max() + 1, 100)
y = boundary_slope * x + boundary_intercept

# ── Step 4: Plot everything ──────────────────────────────────────────────
plt.figure(figsize=(8, 6))

# Scatter plot — each dot coloured by its label
plt.scatter(
    features[:, 0],    # x position of each point
    features[:, 1],    # y position of each point
    c=labels,          # colour = class label (0 or 1)
    cmap='coolwarm',   # blue for 0, red for 1
    edgecolors='k',    # black border on each dot
    s=80               # dot size
)

# Plot the centroids
plt.scatter(*centroid_0, color='blue', marker='X', s=300, edgecolors='black', linewidth=2, label='Centroid 0')
plt.scatter(*centroid_1, color='red', marker='X', s=300, edgecolors='black', linewidth=2, label='Centroid 1')

# Draw the optimal decision boundary
plt.plot(x, y, color='green', linewidth=2, label='Decision Boundary (Optimal)')

plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.title('Binary Classification — Two Clusters with Optimal Separator')
plt.legend()
plt.xlim(features[:, 0].min() - 1, features[:, 0].max() + 1)
plt.ylim(features[:, 1].min() - 1, features[:, 1].max() + 1)
plt.grid(True, alpha=0.3)
plt.show()
````

````python
# ── Testing: Predict which cluster new points belong to ─────────────────
# The decision boundary equation: y = boundary_slope * x + boundary_intercept
# If a point is ABOVE the line → likely cluster 1 (red)
# If a point is BELOW the line → likely cluster 0 (blue)

def predict_cluster(x_point, y_point, slope, intercept):
    """
    Predict which cluster a point belongs to based on decision boundary.
    
    The decision boundary is: y_boundary = slope * x + intercept
    - If y_point > y_boundary → Point is above the line → Class 1
    - If y_point < y_boundary → Point is below the line → Class 0
    """
    y_boundary = slope * x_point + intercept
    
    if y_point > y_boundary:
        return 1, y_boundary
    else:
        return 0, y_boundary

# Test with several new points
test_points = [
    (-8, 8),    # Should be in cluster 0 (blue) - left side
    (0, 0),     # Should be in cluster 1 (red) - right side
    (-5, 5),    # Should be in cluster 0 (blue)
    (8, -2),    # Should be in cluster 1 (red)
]

print("=" * 60)
print("TESTING NEW POINTS ON THE DECISION BOUNDARY")
print("=" * 60)

for x_test, y_test in test_points:
    predicted_cluster, y_line = predict_cluster(x_test, y_test, boundary_slope, boundary_intercept)
    distance = y_test - y_line  # positive = above line (class 1), negative = below line (class 0)
    
    print(f"\nNew Point: ({x_test}, {y_test})")
    print(f"  Boundary y-value at x={x_test}: {y_line:.2f}")
    print(f"  Distance from boundary: {distance:.2f}")
    print(f"  ➜ Predicted Cluster: {predicted_cluster} {'(RED - Cluster 1)' if predicted_cluster == 1 else '(BLUE - Cluster 0)'}")

# ── Visualize the test points on the plot ─────────────────────────────────
plt.figure(figsize=(10, 7))

# Original scatter plot
plt.scatter(
    features[:, 0],
    features[:, 1],
    c=labels,
    cmap='coolwarm',
    edgecolors='k',
    s=80,
    alpha=0.6,
    label='Training Data'
)

# Centroids
plt.scatter(*centroid_0, color='blue', marker='X', s=300, edgecolors='black', linewidth=2, label='Centroid 0')
plt.scatter(*centroid_1, color='red', marker='X', s=300, edgecolors='black', linewidth=2, label='Centroid 1')

# Decision boundary
x_line = np.linspace(features[:, 0].min() - 1, features[:, 0].max() + 1, 100)
y_line = boundary_slope * x_line + boundary_intercept
plt.plot(x_line, y_line, color='green', linewidth=2, label='Decision Boundary')

# Test points
for x_test, y_test in test_points:
    predicted_cluster, _ = predict_cluster(x_test, y_test, boundary_slope, boundary_intercept)
    color = 'red' if predicted_cluster == 1 else 'blue'
    marker_style = 's'  # square marker for test points
    plt.scatter(x_test, y_test, color=color, marker=marker_style, s=200, 
                edgecolors='yellow', linewidth=3, label=f'Test Point ({x_test}, {y_test})')

plt.xlabel('Feature 1', fontsize=12)
plt.ylabel('Feature 2', fontsize=12)
plt.title('Decision Boundary with Test Points', fontsize=14)
plt.legend(loc='best', fontsize=9)
plt.xlim(features[:, 0].min() - 2, features[:, 0].max() + 2)
plt.ylim(features[:, 1].min() - 2, features[:, 1].max() + 2)
plt.grid(True, alpha=0.3)
plt.show()
````

# Understanding GPU Memory vs CPU Memory - Deep Dive

## What does ".numpy() brings data back to CPU" mean?

When you create a TensorFlow tensor, the data can live in **two different places**:

### Memory Hierarchy:

```
┌─────────────────────────────────────────────────────┐
│                  YOUR COMPUTER                       │
├─────────────────────────────────────────────────────┤
│                                                      │
│  ┌──────────────────────┐    ┌──────────────────┐  │
│  │   CPU (Main Processor)   │    │  GPU (Graphics Card) │  │
│  │                          │    │                      │  │
│  │ RAM (Heap Memory)        │    │  VRAM (GPU Memory)   │  │
│  │ • Slower                 │    │  • Much Faster       │  │
│  │ • General purpose        │    │  • Optimized for     │  │
│  │ • Larger capacity        │    │    parallel math     │  │
│  │                          │    │  • Limited size      │  │
│  └──────────────────────────┘    └──────────────────┘  │
│         ↑                                ↑              │
│         └────────────────────────────────┘              │
│         Data transfer via PCIe bus (slow!)             │
│                                                        │
└─────────────────────────────────────────────────────┘
```

## Java Parallelism Analogy

Think of it like **Java threads and memory models**:

```
// JAVA ANALOGY

// TensorFlow tensor is like a thread running on GPU
Thread gpuThread = new Thread(() -> {
    // This thread runs on the GPU (separate processor)
    // It has its own memory space (VRAM)
    // CPU can't directly access this memory
});

// When you call .numpy(), it's like:
// 1. Tell the GPU thread to finish its work
// 2. Copy the result from GPU memory to CPU Heap
// 3. Return it to CPU control

// Example:
int[] result = copyFromGPUToRAM();  // ← This is what .numpy() does!
// Now 'result' is in Java Heap (CPU RAM), accessible by CPU threads
```

## Visual Comparison:

### TensorFlow Tensor (on GPU):
```
hello = tf.constant("Hello!")
↓
Memory Location: VRAM (GPU memory)
Accessible by: GPU only
Speed: ⚡⚡⚡ (very fast for GPU operations)
Access from Python: ❌ (cannot access directly)
```

### NumPy Array (on CPU):
```
hello.numpy()
↓
Memory Location: RAM (CPU Heap memory)
Accessible by: CPU, Python interpreter
Speed: ⚡ (slower for bulk operations, but normal Python)
Access from Python: ✅ (direct access)
```

## Why is GPU Memory Efficient for GPUs?

### GPU Architecture:

```
CPU (1 core - does 1 thing at a time):
   [Process 1] → [Process 2] → [Process 3]
   
GPU (1000s of cores - do things in parallel):
   [1][2][3][4][5]...[1000]
   Processing 1000 things SIMULTANEOUSLY
```

### Why GPU Memory is Optimized:

1. **Parallel Access**: GPU has thousands of cores accessing memory at once
   - CPU Heap: Designed for sequential access (1 thread at a time)
   - GPU VRAM: Designed for parallel access (1000 cores at once)

2. **Memory Bandwidth**:
   - CPU RAM: ~100 GB/s
   - GPU VRAM: ~900 GB/s (9x faster!)

3. **Cache Design**:
   - CPU: Large caches for sequential patterns
   - GPU: Small caches but massive parallel throughput

### Performance Example:

```python
# Matrix multiplication: 1000x1000 matrix

# On CPU (Python/NumPy):
import numpy as np
A = np.random.rand(1000, 1000)
B = np.random.rand(1000, 1000)
result = A @ B  # Takes ~100ms

# On GPU (TensorFlow):
import tensorflow as tf
A_gpu = tf.constant(A)
B_gpu = tf.constant(B)
result_gpu = A_gpu @ B_gpu  # Takes ~5ms (20x faster!)

# Why? GPU processes multiply-add operations in parallel
# 1 core adds 1 element
# 1000 cores add 1000 elements simultaneously
```

## The PCIe Transfer Bottleneck:

```
EXPENSIVE OPERATION - Moving data between CPU and GPU:

GPU Computation (5ms) vs Data Transfer (100ms)

┌─────────────┐              ┌─────────────┐
│   CPU RAM   │              │  GPU VRAM   │
│ [Data A]    │──PCIe Bus────→ [Data A]   │
│ [Data B]    │  (slow!)     │ [Data B]   │
└─────────────┘              └─────────────┘
     ↑                             ↓
     │                        (compute fast)
     └──────────────────────────────┘
              (copy back - slow!)

Best Practice: Keep data on GPU as long as possible!
Don't call .numpy() until you need the result.
```

## When to Use Each:

| Scenario | Use | Why |
|----------|-----|-----|
| **Large matrix operations** | TensorFlow (GPU) | Parallel computation |
| **Python data manipulation** | NumPy (CPU) | CPU-friendly operations |
| **Final result retrieval** | .numpy() (copy to CPU) | Bring result back |
| **Debugging** | .numpy() (copy to CPU) | Print and inspect values |
| **TensorFlow operations** | Keep on GPU | Avoid data transfer |

## Code Example - Efficient vs Inefficient:

```python
import tensorflow as tf

# ❌ INEFFICIENT - Moving data constantly
tensor = tf.constant([1, 2, 3, 4, 5])
result1 = tensor.numpy() + tensor.numpy()  # Copy to CPU twice
result2 = result1 + tensor.numpy()         # Copy again
result3 = result2 + tensor.numpy()         # Copy again
# Total: 4x PCIe transfers!

# ✅ EFFICIENT - Keep on GPU, transfer once at end
tensor = tf.constant([1, 2, 3, 4, 5])
result = tensor + tensor                    # GPU (fast)
result = result + tensor                    # GPU (fast)
result = result + tensor                    # GPU (fast)
final = result.numpy()                      # Copy once to CPU
# Total: 1x PCIe transfer!
```

## Memory Model Summary:

```
Heap Memory (Java/Python analogy):

Java Heap (RAM):
┌─────────────────────────────────────────┐
│ New ArrayList() → occupies heap space   │
│ New Thread() → can access heap          │
│ Garbage collected when unused           │
└─────────────────────────────────────────┘

Python/NumPy Heap (RAM):
┌─────────────────────────────────────────┐
│ np.array([1,2,3]) → occupies RAM        │
│ .numpy() → creates object in Python heap│
│ Garbage collected when unused           │
└─────────────────────────────────────────┘

GPU VRAM (Different Memory - NOT heap):
┌─────────────────────────────────────────┐
│ tf.constant() → occupies GPU VRAM       │
│ tf.Variable() → persistent GPU memory   │
│ Deallocated when TF session ends        │
└─────────────────────────────────────────┘
```

````python
# TensorFlow 1.x vs 2.x - Session Comparison

import tensorflow as tf

print("=" * 70)
print("TensorFlow VERSION:", tf.__version__)
print("=" * 70)

# ─── TensorFlow 1.x APPROACH (Old) ───────────────────────────────────
print("\n❌ TensorFlow 1.x (DEPRECATED):")
print("""
# This NO LONGER WORKS in TensorFlow 2.x:
with tf.Session() as sess:
    hello = tf.constant("Hello, TensorFlow!")
    result = sess.run(hello)
    print(result)
""")

# ─── TensorFlow 2.x APPROACH (New) ──────────────────────────────────
print("\n✅ TensorFlow 2.x (CURRENT):")
print("Eager execution is enabled by default!")
print("No need for sessions - just execute directly:\n")

# Method 1: Direct execution (simplest)
hello = tf.constant("Hello, TensorFlow!")
print(f"Method 1 - Direct: {hello.numpy()}")

# Method 2: With tf.function (for optimization)
@tf.function
def compute_something():
    tensor1 = tf.constant([1, 2, 3])
    tensor2 = tf.constant([4, 5, 6])
    result = tensor1 + tensor2
    return result

result = compute_something()
print(f"Method 2 - tf.function: {result.numpy()}")

# ─── If you REALLY need graph execution behavior ──────────────────────
print("\n📌 If you need TensorFlow 1.x behavior (rare cases):")
print("""
You can use @tf.function decorator to compile operations into a graph.
This is more efficient for repeated operations.
""")

@tf.function
def expensive_computation(x, y):
    '''This gets compiled into a graph for efficiency'''
    z = tf.matmul(x, y)
    return z + 1

# Create sample tensors
A = tf.constant([[1.0, 2.0], [3.0, 4.0]])
B = tf.constant([[2.0, 0.0], [1.0, 2.0]])

result = expensive_computation(A, B)
print(f"Graph execution result:\n{result.numpy()}")

# ─── Summary ────────────────────────────────────────────────────────
print("\n" + "=" * 70)
print("KEY DIFFERENCES:")
print("=" * 70)
print("""
TensorFlow 1.x:
  • Need to create a Session
  • Operations are lazy (don't execute until sess.run())
  • Need to manually manage graphs

TensorFlow 2.x:
  • Eager execution by default (operations run immediately)
  • No sessions needed
  • More Pythonic and easier to debug
  • Use @tf.function for performance optimization
""")
````

# TensorFlow Variables & Initialization - Why Variables Need Init in 1.x

## What's the Difference: Constant vs Variable?

### tf.constant()
```python
x = tf.constant(5)  # ← Fixed value, cannot change
y = tf.constant([1, 2, 3])  # ← Static data
```
Constants are **immutable** - once created, they never change. No initialization needed.

### tf.Variable()
```python
w = tf.Variable(initial_value=5)  # ← Can change over time
b = tf.Variable(tf.zeros([128]))  # ← Mutable state
```
Variables are **mutable** - their values can be updated. This is where **initialization** becomes important.

## Why Variables Need Initialization

Think of variables like **uninitialized memory slots**:

```
┌──────────────────────────────────────────────────┐
│         Computer Memory (RAM/VRAM)                │
├──────────────────────────────────────────────────┤
│                                                  │
│  Memory Address: 0x1000                         │
│  ┌─────────────────────────────────────┐        │
│  │ Variable w: [?????? ?????? ??????]  │ ← Garbage data!
│  │ (Not yet initialized)                │        │
│  └─────────────────────────────────────┘        │
│                                                  │
│  If you try to use this before init:            │
│  ❌ You get random/undefined values             │
│  ❌ Your neural network is broken               │
│                                                  │
└──────────────────────────────────────────────────┘
```

## TensorFlow 1.x: Initialization Required

In TensorFlow 1.x, variables were **lazy** - they existed in the graph but had no actual values until initialized:

```python
# TensorFlow 1.x - MUST initialize

import tensorflow as tf

# Create variables
W = tf.Variable(tf.random_normal([784, 128]))  # Weight matrix
b = tf.Variable(tf.zeros([128]))               # Bias vector

# Define operations using these variables
output = tf.matmul(input_data, W) + b

# Create a session
with tf.Session() as sess:
    # CRITICAL: Initialize ALL variables before using them
    sess.run(tf.global_variables_initializer())  # ← THIS IS REQUIRED!
    
    # NOW you can run operations
    result = sess.run(output, feed_dict={input_data: X})
    print(result)  # ✅ Works because variables are initialized

# If you forgot the initialization line:
# with tf.Session() as sess:
#     result = sess.run(output, feed_dict={input_data: X})
#     # ❌ ERROR: Variable w:0 has not been initialized
```

### The Two-Phase Process in TensorFlow 1.x:

```
Phase 1: Graph Building (Python code runs)
┌─────────────────────────────────────────────┐
│ W = tf.Variable(tf.random_normal([...]))   │
│ b = tf.Variable(tf.zeros([...]))           │
│ output = tf.matmul(input, W) + b           │
│                                             │
│ ← Variables exist in the GRAPH but have    │
│   no actual numerical values yet!          │
└─────────────────────────────────────────────┘
                    ↓
Phase 2: Session Execution
┌─────────────────────────────────────────────┐
│ with tf.Session() as sess:                 │
│     sess.run(tf.global_variables_init())   │
│     # ← NOW allocates memory and sets      │
│     #   random_normal values into W        │
│     #   zeros into b                       │
│                                             │
│     result = sess.run(output)  ✅ Works!   │
└─────────────────────────────────────────────┘
```

## Detailed Example: What Happens Step-by-Step

### Step 1: Create Variables (Graph Building)

```python
import tensorflow as tf

W = tf.Variable(tf.random_normal([3, 2]))
print(W)  # Output: <tf.Variable 'Variable:0' shape=(3, 2) dtype=float32>
          # ← Just a description! No actual values yet!
```

### Step 2: Try to Use Without Initialization (❌ FAILS)

```python
with tf.Session() as sess:
    print(sess.run(W))  # ❌ ERROR!
    # RuntimeError: Attempting to capture an EagerTensor without building a function.
    # OR: FailedPreconditionError: Attempting to use uninitialized value Variable
```

### Step 3: Initialize and Use (✅ WORKS)

```python
with tf.Session() as sess:
    sess.run(tf.global_variables_initializer())  # Initialize ALL variables
    print(sess.run(W))  # ✅ Now it prints actual values!
    # Output something like:
    # [[ 0.45  -0.23]
    #  [-1.12   0.89]
    #  [ 0.34  -0.56]]
```

## What Does tf.global_variables_initializer() Do?

```python
initializer = tf.global_variables_initializer()
```

This creates an **operation** that, when run, will:

1. Find ALL variables in the graph
2. Allocate memory for each variable
3. Call their initialization functions (random_normal, zeros, etc.)
4. Assign the resulting values to the memory

```
Initialization Process:

┌────────────────────────────────────────────────────────┐
│  Variables in Graph:                                   │
│  • W = tf.Variable(tf.random_normal(...))             │
│  • b = tf.Variable(tf.zeros(...))                     │
│  • weights = tf.Variable(tf.truncated_normal(...))    │
└────────────────────────────────────────────────────────┘
                         ↓
            sess.run(global_variables_init)
                         ↓
┌────────────────────────────────────────────────────────┐
│  For each variable:                                    │
│  1. Allocate memory space                             │
│  2. Run the initializer function                      │
│     • random_normal → generate random values         │
│     • zeros → fill with 0.0                          │
│     • truncated_normal → generate clipped values     │
│  3. Copy values into the memory space                 │
│  4. Variable is now ready to use                      │
└────────────────────────────────────────────────────────┘
                         ↓
            Variables have actual numerical values
            Ready for computation!
```

## You Can Initialize Variables Individually Too

```python
import tensorflow as tf

W = tf.Variable(tf.random_normal([3, 2]))
b = tf.Variable(tf.zeros([2]))

with tf.Session() as sess:
    # Option 1: Initialize ALL at once
    sess.run(tf.global_variables_initializer())
    
    # Option 2: Initialize individually
    sess.run(W.initializer)  # Initialize just W
    sess.run(b.initializer)  # Initialize just b
    
    # Option 3: List multiple
    sess.run([W.initializer, b.initializer])
```

## Common Initialization Methods

```python
import tensorflow as tf

b = tf.Variable(tf.zeros([128]))
# → Creates all zeros: [0, 0, 0, ..., 0]

# 6. Constant initialization

# → Fixed value: 5.0
```

## TensorFlow 2.x: No Initialization Needed!

This is one reason TensorFlow 2.x is simpler:

```python
# TensorFlow 2.x - Eager execution, automatic initialization

import tensorflow as tf

# Create variables
W = tf.Variable(tf.random.normal([784, 128]))
b = tf.Variable(tf.zeros([128]))

# Use them immediately! No initialization needed!
print(W)  # Prints actual values, not just description
print(b)  # Prints [0. 0. 0. ...]

# Forward pass
output = tf.matmul(input_data, W) + b
print(output)  # ✅ Works immediately!
```

**Why the difference?**
- **TensorFlow 1.x**: Lazy evaluation → build graph first → then execute with session → then initialize
- **TensorFlow 2.x**: Eager execution → operations run immediately → initialization happens automatically

## Key Takeaway

**Variables in TensorFlow 1.x are like blank slots in memory** - they need to be filled with initial values before you can use them. The `initializer` operation tells TensorFlow what to put in those slots (random values, zeros, etc.), and `sess.run(global_variables_initializer())` actually performs that initialization.

In TensorFlow 2.x, this happens automatically and invisibly, making the code much simpler.

# TensorFlow 2.x: Variables & Initialization (Automatic & Simplified)

## The Big Difference: Automatic Initialization

In TensorFlow 2.x, variable initialization happens **automatically and invisibly**. You don't need to think about it:

```python
# TensorFlow 2.x - Just works!

import tensorflow as tf

# Create a variable
W = tf.Variable(tf.random.normal([3, 2]))

# Print it immediately - NO initialization needed!
print(W)
# Output: <tf.Variable 'Variable:0' shape=(3, 2) dtype=float32, numpy=
# [[ 0.45  -0.23]
#  [-1.12   0.89]
#  [ 0.34  -0.56]]>

# Use it immediately
result = W * 2
print(result)  # ✅ Works right away!
```

## Why TensorFlow 2.x Doesn't Need Initialization

TensorFlow 2.x uses **eager execution** instead of lazy graph building:

```
TensorFlow 1.x (Lazy):
┌─────────────────────────────┐
│ 1. Define variables         │ ← Creates graph nodes (no values)
│ 2. Build computation graph  │ ← Still no values
│ 3. Create session           │ ← Still no values
│ 4. Initialize              │ ← NOW allocate memory & fill values
│ 5. Run operations          │ ← NOW execute with actual values
└─────────────────────────────┘

TensorFlow 2.x (Eager):
┌─────────────────────────────┐
│ 1. Create variable          │ ← Immediately allocates memory
│ 2. Fill with initial values │ ← Automatically done
│ 3. Use immediately         │ ← Ready to go!
└─────────────────────────────┘
```

When you create a variable, TensorFlow 2.x:

```

│            [ 0.34, -0.56]]          │
└──────────────────────────────────────┘
              ↓
Step 3: Allocate memory and store values
┌──────────────────────────────────────┐
│ Memory Address: 0x1000              │
│ [0.45] [-0.23] [-1.12] [0.89] ... │
│                                     │

              ↓

# Deep Dive: TensorFlow Sessions - Concept, Architecture & Why They're Gone

## What is a TensorFlow Session?

A **session** in TensorFlow 1.x was a **runtime environment** that executed operations. Think of it like a **virtual machine inside your program**:

```
TensorFlow 1.x Architecture:

┌─────────────────────────────────────────────────────┐
│  Your Python Code                                   │
├─────────────────────────────────────────────────────┤
│                                                      │
│  x = tf.constant(5)                                │
│  y = tf.constant(3)                                │
│  z = x + y  ← This DOESN'T execute yet!            │
│              It builds a GRAPH                      │
│                                                      │
└───────────┬──────────────────────────────────────────┘
            │
            ↓
┌─────────────────────────────────────────────────────┐
│  TensorFlow Graph (Blueprint/Plan)                  │
├─────────────────────────────────────────────────────┤
│                                                      │
│      ┌─────────┐     ┌─────────┐                   │
│      │ Const 5 │────→│  Add    │────→ [Result]   │
│      └─────────┘     ↑         │                   │

## Key Concept: Lazy Evaluation

In TensorFlow 1.x, operations were **lazy**:

```python
# TensorFlow 1.x (Lazy Evaluation)

x = tf.constant(5)
y = tf.constant(3)
z = x + y

print(z)  # Output: Tensor("add:0", shape=(), dtype=int32)
          # ← NOT the actual value! Just the tensor description

# To get the actual value, you NEED a session:
with tf.Session() as sess:
    result = sess.run(z)  # Now it computes
    print(result)         # Output: 8
```

### Why was this design used?

1. **Graph Optimization**: The entire computation graph could be analyzed before execution
2. **Distributed Computing**: The graph could be sent to different devices (GPUs, TPUs, multiple machines)
3. **Memory Efficiency**: Operations could be optimized and fused together
4. **Hardware Compatibility**: The session handled device placement automatically

```
Optimization Example:

Graph without session:
    [Add] → [Multiply] → [Add] → [Result]

Graph optimizer (inside session):
    Could reorder, fuse, or eliminate operations for efficiency
    [Add+Multiply+Add optimized] → [Result]  (faster!)
```

## The Problem: Verbosity & Complexity

While graphs were powerful, the session model was **cumbersome** for users:

```python
# TensorFlow 1.x - Very verbose

# Build the graph

```

## Why Sessions Were Removed (Technical Reasons)

### Reason 1: Eager Execution Simplifies Debugging

```python
# TensorFlow 1.x - Can't debug intermediate values
with tf.Session() as sess:
    x = tf.constant([1, 2, 3])
    y = tf.sqrt(x)
    z = y * 2
    result = sess.run(z)
    
    # If result is wrong, you can't print y to debug!

# TensorFlow 2.x - Debug easily!
x = tf.constant([1, 2, 3])
y = tf.sqrt(x)
print(y)  # ← Can print intermediate values!
z = y * 2
print(z)  # ← See each step clearly
```

### Reason 2: Python Integration

```python
# TensorFlow 1.x - NOT Pythonic
def complex_computation(x_val):
    x = tf.placeholder(tf.float32)
    y = x * 2
    z = y + 5
    
    with tf.Session() as sess:
        return sess.run(z, feed_dict={x: x_val})

# TensorFlow 2.x - Pure Python
def complex_computation(x_val):
    x = tf.constant(x_val, dtype=tf.float32)
    y = x * 2
    z = y + 5
    return z.numpy()  # Convert to NumPy when needed
```

### Reason 3: Keras Integration

Google wanted TensorFlow to be the backend for **Keras** (high-level ML library). Keras demanded a simpler, more Pythonic interface.

```
TensorFlow 1.x + Keras:
    Complex, dual APIs, confusing

TensorFlow 2.x = TensorFlow + Keras merged:
    Single, simple, unified API
```

### Reason 4: Auto Differentiation & @tf.function

When you need performance, you don't need full sessions - just the `@tf.function` decorator:

```python
# TensorFlow 2.x - Get graph benefits without sessions!

@tf.function  # This creates an optimized graph
def expensive_operation(x, y):
    z = tf.matmul(x, y)
    return z + 1

# First call: Python code is traced and converted to graph
# Subsequent calls: Fast graph execution

A = tf.constant([[1.0, 2.0], [3.0, 4.0]])
B = tf.constant([[2.0, 0.0], [1.0, 2.0]])

result = expensive_operation(A, B)  # Fast!
```

# Understanding Centroids and Decision Boundaries

## Step 1: What is `make_blobs`?

Yes! `make_blobs` creates random points, but **NOT completely random**. It creates points in **2 clusters** around 2 center points:

```
make_blobs with centers=2 creates:
- Cluster 0: 25 random points clustered around CENTER 0 (e.g., around point (-9, 5))
- Cluster 1: 25 random points clustered around CENTER 1 (e.g., around point (8, -2))
```

### Visual Example:
```
Cluster 0 (Blue)              Cluster 1 (Red)
  •  •  •                           •  •  •
    • • •                         •   •   •
  •  •  •  (centered somewhere)  •  •  •  (centered somewhere else)
```

## Step 2: What is a Centroid?

A **centroid** is simply the **average position** of all points in a cluster.

### Formula:
$$\text{Centroid} = \frac{\sum \text{all points in cluster}}{n}$$

For 2D points (x, y):
```
Centroid_x = (x₁ + x₂ + x₃ + ... + xₙ) / n
Centroid_y = (y₁ + y₂ + y₃ + ... + yₙ) / n
```

## Step 3: How we calculated it in the code

```python
centroid_0 = features[labels == 0].mean(axis=0)
centroid_1 = features[labels == 1].mean(axis=0)
```

### Breaking this down:

| Part | Meaning |
|------|---------|
| `features[labels == 0]` | Select all points where label is 0 (blue cluster) |
| `.mean(axis=0)` | Calculate the average of each column (x and y coordinates) |
| `centroid_0` | The center point of cluster 0 |

### Example:
```python
# If cluster 0 has these points:
features[labels == 0] = [[-8.5, 6.2],
                         [-7.2, 5.8],
                         [-9.1, 5.5],
                         ...]

# The centroid is the average:
centroid_0 = [mean of all x], [mean of all y]
centroid_0 = [-8.3, 5.7]  # The center of all these blue points
```

## Step 4: Why use centroids?

The centroid is the **perfect representative** of each cluster because it's the point **closest to all other points in that cluster** on average.

### The Decision Boundary Logic:

1. **Find centroid of cluster 0** → Center of blue points
2. **Find centroid of cluster 1** → Center of red points
3. **Draw a line perpendicular to the line connecting the two centroids**
4. **Pass this line through the midpoint** between the two centroids

This creates a **linear classifier** that separates the two clusters! 🎯

## Visual Summary:

```
Blue Cluster            Red Cluster
   •                       •
  • •          MIDPOINT   • •
 • • •  -------- X -------• • •
  • •    Centroids        • •
   •      Distance         •
   
   ↓
   
Decision Boundary (perpendicular line through midpoint):
        ╱
       ╱  ← Separates blue from red
      ╱
```

# Training a Simple Linear Model in TensorFlow (Low-Level API)

This example demonstrates how to:
- Define input data, weights, and outputs
- Build a linear model (like a single neuron)
- Train the model using gradient descent (without Keras)
- See how the weights and bias are updated during training

**Key steps:**
1. Prepare input and output data
2. Initialize weights and bias as trainable variables
3. Define the model and loss function
4. Use an optimizer to minimize the loss
5. Run a training loop and observe learning
6. Test the trained model on new data


````python
import tensorflow as tf
import numpy as np

# 1. Define input data (X) and target output (Y)
X = np.array([[1.0], [2.0], [3.0], [4.0]], dtype=np.float32)  # shape (4, 1)
Y = np.array([[2.0], [4.0], [6.0], [8.0]], dtype=np.float32)  # shape (4, 1)

# 2. Define weights and bias as trainable variables
W = tf.Variable(tf.random.normal([1, 2]))  # shape (input_dim, output_dim)
b = tf.Variable(tf.zeros([1]))

# 3. Define the model (linear: y = X*W + b)
def model(x):
    return tf.matmul(x, W) + b

# 4. Define the loss function (Mean Squared Error)
def loss_fn(y_true, y_pred):
    return tf.reduce_mean(tf.square(y_true - y_pred))

# 5. Define the optimizer
optimizer = tf.optimizers.SGD(learning_rate=0.1)

# 6. Training loop
for epoch in range(200):
    with tf.GradientTape() as tape:
        y_pred = model(X)
        loss = loss_fn(Y, y_pred)
    gradients = tape.gradient(loss, [W, b])
    optimizer.apply_gradients(zip(gradients, [W, b]))
    if (epoch+1) % 20 == 0:
        print(f"Epoch {epoch+1}: Loss = {loss.numpy():.4f}, W = {W.numpy().flatten()}, b = {b.numpy()}")

# 7. Test the trained model
test_X = np.array([[5.0]], dtype=np.float32)
pred = model(test_X)
print(f"Prediction for input 5.0: {pred.numpy().flatten()[0]:.2f}")
````

````python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

# Generate noisy data
x = np.linspace(-10, 10, 10) + np.random.uniform(-1.5, 1.5, 10)
y = np.linspace(-20, 20, 10) + np.random.uniform(-1.5, 1.5, 10)

# Plot the data
plt.plot(x, y, '*', label='Data Points')

# Initialize variables for slope (m) and intercept (b)
randoms = np.random.rand(2)
m = tf.Variable(randoms[0], dtype=tf.float32)
b = tf.Variable(randoms[1], dtype=tf.float32)
print(f"Initial random slope (m): {m.numpy():.4f}")
print(f"Initial random intercept (b): {b.numpy():.4f}")

# Convert x and y to tensors
x_tensor = tf.constant(x, dtype=tf.float32)
y_tensor = tf.constant(y, dtype=tf.float32)

# Define optimizer
optimizer = tf.keras.optimizers.SGD(learning_rate=0.01)

# Training loop
for epoch in range(500):
    with tf.GradientTape() as tape:
        y_pred = m * x_tensor + b
        loss = tf.reduce_mean((y_tensor - y_pred) ** 2)
    gradients = tape.gradient(loss, [m, b])
    optimizer.apply_gradients(zip(gradients, [m, b]))
    if (epoch+1) % 100 == 0:
        print(f"Epoch {epoch+1}: Loss = {loss.numpy():.4f}, m = {m.numpy():.4f}, b = {b.numpy():.4f}")

# Final fitted line
y_fit = m.numpy() * x + b.numpy()
plt.plot(x, y_fit, 'r-', label='Fitted Line')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Linear Regression Fit (TensorFlow)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
````

### Explanation of the Training Loop

- `for epoch in range(200):`
  - Runs the training loop for 200 epochs (iterations).

- `with tf.GradientTape() as tape:`
  - Records all operations inside the block for automatic differentiation (so TensorFlow can compute gradients).

- `y_pred = model(X)`
  - Computes the model’s predictions for the input data `X` using the current weights and bias.

- `loss = loss_fn(Y, y_pred)`
  - Calculates the loss (error) between the true outputs `Y` and the predicted outputs `y_pred`.

- `gradients = tape.gradient(loss, [W, b])`
  - Computes the gradients (partial derivatives) of the loss with respect to the variables `[W, b]`.
  - **Why `[W, b]`?**
    - You want to know how much to change each parameter (W and b) to reduce the loss.
    - Passing `[W, b]` tells TensorFlow to compute gradients for both variables.

- `optimizer.apply_gradients(zip(gradients, [W, b]))`
  - Updates the variables `[W, b]` using the computed gradients and the optimizer (e.g., SGD).
  - `zip(gradients, [W, b])` pairs each gradient with its corresponding variable.

- `if (epoch+1) % 20 == 0:`
  - Every 20 epochs, print the current loss to monitor training progress.

**Summary:**
- You pass `[W, b]` to `tape.gradient` so TensorFlow knows which variables to compute gradients for and update during training.
- This is how the model learns: by adjusting `W` and `b` to minimize the loss.
